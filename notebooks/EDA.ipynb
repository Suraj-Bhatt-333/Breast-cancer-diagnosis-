```python
#  droping id because it does not provide any medical information to us 
df = df.drop("id", axis=1)
```

```python
df.head(2)
```

```python
#checking for the data distribution ( whether it is imbalanced or not )
df["diagnosis"].value_counts()
```

```python
df["diagnosis"].value_counts(normalize=True) * 100
```

```python

df["diagnosis"].value_counts().plot(kind="bar")

plt.title("Diagnosis Distribution")
plt.xlabel("Diagnosis")
plt.ylabel("Count")
plt.show()
```

```python
# as data distribution is fine we dont have to apply here undersampling or SMOTE, lets move forward 
```

```python
#encoding the target columns
df["diagnosis"] = df["diagnosis"].map({
    "B": 0,
    "M": 1
})
```

```python
df["diagnosis"].value_counts()
```

```python
#correlation heatmap
corr = df.corr()

plt.figure(figsize=(15,12))
plt.imshow(corr, cmap="coolwarm")
plt.colorbar()

plt.xticks(range(len(corr.columns)), corr.columns, rotation=90, fontsize=7)
plt.yticks(range(len(corr.columns)), corr.columns, fontsize=7)

plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

```python
important_features = [
    "radius_mean",
    "texture_mean",
    "perimeter_mean",
    "area_mean",
    "concavity_mean"
]

for feature in important_features:
    plt.figure(figsize=(6,3))
    plt.hist(df[feature], bins=30)
    plt.title(feature)
    plt.xlabel(feature)
    plt.ylabel("Frequency")
    plt.show()
```

```python
for feature in important_features:
    plt.figure(figsize=(6,2))
    plt.boxplot(df[feature], vert=False)
    plt.title(feature)
    plt.show()
```
